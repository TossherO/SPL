```python
import os
import pickle
import numpy as np
from nuscenes.nuscenes import NuScenes
```

```python
data_root = '../data/nuscenes'
info_path = '../data/nuscenes/nuscenes_data_info.pkl'
nusc = NuScenes(version='v1.0-trainval-select', dataroot=data_root, verbose=False)
with open(info_path, 'rb') as f:
    data_info = pickle.load(f)
classes = {0: 'person', 1: 'bicycle', 2: 'car', 3: 'motorcycle', 5: 'bus', 7: 'truck'}
cams = data_info[0]['samples'][0]['cams'].keys()
```

```python
scene_idx = 0
idx = 15
scene_save_dir = os.path.join(data_root, '3d_objects', data_info[scene_idx]['scene_name'])
sample = data_info[scene_idx]['samples'][idx]
results_path = os.path.join(scene_save_dir, os.path.basename(sample['lidar_path']).replace('.bin', '.pkl'))
with open(results_path, 'rb') as f:
    results = pickle.load(f)
```

```python
num_points_dict = {
    'person': [],
    'bicycle': [],
    'car': [],
    'motorcycle': [],
    'bus': [],
    'truck': []
}

for scene_idx in range(len(data_info)):
    scene_save_dir = os.path.join(data_root, '3d_objects', data_info[scene_idx]['scene_name'])
    for idx in range(len(data_info[scene_idx]['samples'])):
        sample = data_info[scene_idx]['samples'][idx]
        results_path = os.path.join(scene_save_dir, os.path.basename(sample['lidar_path']).replace('.bin', '.pkl'))
        with open(results_path, 'rb') as f:
            results = pickle.load(f)
        for obj in results['objects']:
            name = obj['name'].split('_')[0]
            if obj.get('points') is not None:
                num_points = obj['points'].shape[0]
            else:
                num_points = 0
            num_points_dict[name].append(num_points)

for name in num_points_dict:
    num_points = np.array(num_points_dict[name])
    print(f'{name}: {num_points.mean():.2f} ({num_points.std():.2f})')
    print(f'  {name} == 0: {np.sum(num_points == 0)}')
    print(f'  0 < {name} <= 5: {np.sum((num_points > 0) & (num_points <= 5))}')
    print(f'  5 < {name} <= 10: {np.sum((num_points > 5) & (num_points <= 10))}')
    print(f'  10 < {name} <= 20: {np.sum((num_points > 10) & (num_points <= 20))}')
    print(f'  20 < {name} <= 50: {np.sum((num_points > 20) & (num_points <= 50))}')
    print(f'  {name} > 50: {np.sum(num_points > 50)}')
```

```python
labels_count = {}
for scene_idx in range(len(data_info)):
    for idx in range(len(data_info[scene_idx]['samples'])):
        sample = data_info[scene_idx]['samples'][idx]
        for obj in sample['labels']:
            name = obj['name']
            if name not in labels_count:
                labels_count[name] = 0
            labels_count[name] += 1
            
for name in labels_count:
    print(f'{name}: {labels_count[name]}')
```

```python
for name in num_points_dict:
    num_points = np.array(num_points_dict[name])
    print(f'{name}: {np.sum(num_points > 10)}')
```

```python
results_count = {}
for scene_idx in range(len(data_info)):
    scene_save_dir = os.path.join(data_root, '3d_objects', data_info[scene_idx]['scene_name'])
    for idx in range(len(data_info[scene_idx]['samples'])):
        sample = data_info[scene_idx]['samples'][idx]
        results_path = os.path.join(scene_save_dir, os.path.basename(sample['lidar_path']).replace('.bin', '.pkl'))
        with open(results_path, 'rb') as f:
            results = pickle.load(f)
        for obj in results['objects']:
            name = obj['name']
            if name not in results_count:
                results_count[name] = 0
            results_count[name] += 1

for name in results_count:
    print(f'{name}: {results_count[name]}')
```
